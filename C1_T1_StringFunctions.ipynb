#### casefold() - Used for case insensitive comparision/Strong lowercase, also it not only converts uppercase letters to lowercase but also handles special characters and locale-specific cases. Eg- it converts the German letter ß to ss, which lower() does not do.
#### lower() - convert all uppercase letters to lowercase

```python
Client_LaptopName = "HaẞE"
print(Client_LaptopName)
print(Client_LaptopName.casefold())
```

```python
Text = "Claẞ"
print(Text.casefold())
```

#### casefold() VS lower() - 


```python
print(Client_LaptopName.lower() == Client_LaptopName.casefold())
print(Client_LaptopName.lower())
```

```python
Text_1 = "Eẞential"
Text_2 = "ESSENTIAL"
print(Text_1.casefold() == Text_2.casefold())
print(Text_1,Text_2)
print(Text_1.casefold(),Text_2.casefold())
```

```python
Text_1 = "Eẞential"
Text_2 = "ESSENTIAL"
print(Text_1.lower() == Text_2.lower())
print(Text_1,Text_2)
print(Text_1.lower(),Text_2.lower())
```

#### expandtabs() - Used for expanding the tab characters
#### partition(), rpartition() - Partitioning the string
#### removeprefix(), removesuffix() - Used for removing the Begining/Ending Text
#### strip(), lstrip(), rstrip() - Used for removing the spaces/characters


```python
Student_Subject = "Social\tScience"
Student_Name = "Mr. Jeeva"
Student_LName = "Jeeva JR"
#Expandtabs
print("\nBelow is the example for expandtabs")
print(Student_Subject)
print(Student_Subject.expandtabs(10))

##partition
print("\nBelow is the example for partiton")
print(Client_Email.partition('.'))
print(Client_Email.partition('@'))

# rpartition
print("\nBelow is the example for rpartiton")
print(Client_Email.rpartition('.'))
print(Client_Address.rpartition(','))

## removeprefix, strip, lstrip
print("\nBelow is the example for removeprefix")
print(Student_Subject.removeprefix('Social'))
print(Student_Subject.removeprefix('Social').strip())
print(Student_Subject.removeprefix('Social').lstrip())
print(Student_Name.removeprefix('Mr.').lstrip())

## removesuffix
print("\nBelow is the example for removesuffix")
print(Student_Subject.removesuffix('ce'))
print(Student_LName.removesuffix('JR'))

#strip, rstrip, lstrip
print('\n',Client_Address)
print(Client_Email.strip('abc'))
print(Client_Address.lstrip())
print(Client_Address.rstrip())
```

#### isalpha()   - checks it has only alphabets
#### isdigit()   - checks it has only digits
#### isnumeric() - check it has numeric values
#### isascii()   - check whether all characters in the string are ASCII character
#### isdecimal() - check it has decimal characters like those that can be used to form a number in base 10.
#### isalnum()   - check it has alphabets/numeric characters
#### isspace()   - check it has spaces
#### islower()   - check it has full lowercase string
#### isupper()   - check it has full uppercase string
#### istitle()   - checks the starting character of each word is capitalized
#### For all of the above functions the return type is boolean

```python
print('123abc'.isalpha())
print('Aabc'.isalpha())

print('123abc'.isdigit())
print('134'.isdigit())

print('123abc'.isnumeric())
print('134'.isnumeric())
print('134.23'.isnumeric())

print('123abc'.isascii())
print('134'.isascii())
print('134@23'.isascii())
print('blué'.isascii())

print('123abc'.isdecimal())
print('134'.isdecimal())
print('134.23'.isdecimal())
print('65'.isdecimal())

print('123abc'.isalnum())
print('134'.isalnum())
print('123abc@'.isalnum())

print('123abc 456def'.isspace())
print('123abc456def'.isspace())
print(''.isspace())
print(' '.isspace())

print("JeeVa".islower())
print("jeeva".islower())

print("JeeVa".isupper())
print("JEEVA".isupper())

print("Jeeva is a Good Boy".istitle())
print("Jeeva Is A Good Boy".istitle())
```

#### maketrans() - It creates a mapping table for the character replacement
#### translate() - It applies that mapping to a string

```python
Alphabet_1 = "abc"
Alphabet_2 = "def"
Translation = str.maketrans({'a':'1','b':'2','c':'3','d':'4','e':'5','f':'6'})
print(Alphabet_1.translate(Translation)+Alphabet_2.translate(Translation))
```

```python
Alphabet_1 = "abc"
Alphabet_2 = "def"
Translation = str.maketrans('abcdef','123456')
print(Alphabet_1.translate(Translation)+Alphabet_2.translate(Translation))
```

```python
Translation = str.maketrans('abcdefghi','123456789')
print(Client_FirstName.translate(Translation))
```

#### split() - It converts string into list
#### rsplit() - It splits from the right
#### splitlines() - It is used for spliting the string based on lines
#### join() - It is used to concatenate all the elements of an iterables containing strings, with a specified separator between each element.
#### replace() - It is used for replacing part of string
#### find() - It finds the position of the word
#### rfind() - It finds the position of the word from right
#### index() - It returns the index value of the specified word
#### rindex() - IT is similar to index(), but searches from the right.

```python
print("Jeeva is a good boy".split())
print("Jeeva@goodboy.com".split('@'))
print("Jeeva@goodboy.com".split('.'))

print("Jeeva.is.goodboy.com".rsplit('.',1))
print("Jeeva.is.goodboy.com".rsplit('.',2))
print("a,b,c,d,e,f,g,h,i,j".rsplit(',',5))

print("Jeeva\ngoodboy\ncom".splitlines())
print("Jeeva.goodboy\ncom".splitlines())
print("Jeeva\ngoodboy\ncom".splitlines())
print("a\nb\nc\nd,e,f,g,h\ni,j".splitlines())

Student_Subjects_1 = ["Tamil","English","Maths"]
print('.'.join(Student_Subjects_1))
print(' '.join(Student_Subjects_1))
print('@'.join(["abc","abc.com"]))

print(Client_Address.strip())
print(Client_Address.strip().replace('anna','indra'))
fruits = "apple banana jackfruit banana jackfruit pineapple banana"
print(fruits.replace('banana','orange'))
print(fruits.replace('banana','orange',2))
print(fruits.replace('banana','orange',3))

print(fruits)
print(fruits.find("banana"))
print(fruits.rfind("banana"))
print(fruits.index("banana"))
print(fruits.rindex("banana"))
print(fruits.index("jackfruit"))
print(fruits.rindex("jackfruit"))

## find vs index
try:
    print('\n',fruits.rfind("orange"))
    print(fruits.index("orange"))
    

except Exception as e:
    print(e)
```

#### isprintable() - Check whether all the characters in the string is printable

```python
Stu_Name = "Jeeva"
Stu_Email = "Jeeva@Jeev.com"
Stu_Address = "\nStreet No.1 Anna nagar"
print(Stu_Name.isprintable())
print(Stu_Email.isprintable())
print(Stu_Address.isprintable())
```

```python
Texts = [Stu_Name, Stu_Email, Stu_Address]
print(Texts)
for Text in Texts:
    if Text.isprintable():
        print(Text+ " is printable")

    else:
        print(Text + " is not printable")
```

#### Getting input word from the user and counting the vowels on that word

```python
User_Input = input("Please enter the word")
User_Input_length = len(User_Input)
#print(User_Input_length)
vowels_count = 0

for i in range(User_Input_length):
    if User_Input[i].lower() in ['a','e','i','o','u']:
        vowels_count += 1
if vowels_count != 0:       
    print("The Vowels count for the given word {} is {}".format(User_Input,vowels_count))
else:
    print("Sorry! the given word {} has no vowels on it".format(User_Input))
```
