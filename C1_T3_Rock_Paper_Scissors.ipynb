## Task 3
Rock, Paper, Scissors Game
Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer.
This task will test your understanding of:
●	Input/output
●	Conditional logic
●	Random module
●	Loops
●	Exception handling

Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.



```python
import random

print("-----Rock, Paper, Scissors Game-----\n")

loop = True

Computer_Input = ["Rock", "Paper", "Scissors"]
Computer_Input_End = len(Computer_Input)-1

while loop:
    try:
        User_Choice = input("\nPlease enter your choice [Rock, Paper, Scissors] - ").lower()
        Computer_Choice = Computer_Input[random.randint(0,Computer_Input_End)].lower()
        
        if User_Choice == Computer_Choice:
            print("User Choice is {}, Computer Choice is {}, Both are tie!!".format(User_Choice,Computer_Choice))
    
        elif User_Choice == 'rock' and Computer_Choice == 'paper':
            print("User Choice is {}, Computer Choice is {}, Computer Wins!!".format(User_Choice,Computer_Choice))
    
        elif User_Choice == 'rock' and Computer_Choice == 'scissors':
            print("User Choice is {}, Computer Choice is {}, You Wins!!".format(User_Choice,Computer_Choice))
    
        elif User_Choice == 'paper' and Computer_Choice == 'rock':
            print("User Choice is {}, Computer Choice is {}, You Wins!!".format(User_Choice,Computer_Choice))
    
        elif User_Choice == 'paper' and Computer_Choice == 'scissors':
            print("User Choice is {}, Computer Choice is {}, Computer Wins!!".format(User_Choice,Computer_Choice))
    
        elif User_Choice == 'scissors' and Computer_Choice == 'rock':
            print("User Choice is {}, Computer Choice is {}, Computer Wins!!".format(User_Choice,Computer_Choice))
    
        elif User_Choice == 'scissors' and Computer_Choice == 'paper':
            print("User Choice is {}, Computer Choice is {}, You Wins!!".format(User_Choice,Computer_Choice))

        else:    
            raise Exception()
    
        if loop:
            Play_Again = input("\nPlease enter 'yes' if you want to play again or else enter 'No' - ")
            if Play_Again.lower() not in ['yes','y']:
                loop = False
                print("Thanks for playing the rock, paper, scissors game!")
                
    except Exception:
        print("\nInvalid Choice!! Please enter rock, paper, or scissors")
```
