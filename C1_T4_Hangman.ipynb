## Task 4
Word Guessing Game (Mini Hangman)
Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word.

This will test your:
●	Loops
●	Conditional logic
●	String manipulation
●	Exception handling
●	Logical thinking

Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.


```python
import random

print("-----Word Guessing Game (Mini Hangman)-----\n")

loop = True

PreDefined_Input_Words = ["Pine", "Orange", "Rock", "rose", "Python", "bash"]
PreDefined_Input_LenEnd = len(PreDefined_Input_Words)-1

while loop:     
    Random_Word = PreDefined_Input_Words[random.randint(0,PreDefined_Input_LenEnd)].lower()
    print("Random word is " + Random_Word)
    Random_Word_len = len(Random_Word)
    Word = ["_"] * Random_Word_len
    
    Attempt = 6
    
    print("The word has {} Letters, you can guess one letter at a time".format(Random_Word_len))
    
    while Attempt > 0 and "_" in Word:
        try:
            print("\nWord is ( {} ) You have {} Attempts left".format(" ".join(Word), Attempt))
            User_Input = input("\nGuess the letter".lower())

            if len(User_Input) != 1 or not User_Input.isalpha():
                raise Exception()
    
            if User_Input in Random_Word:
                print("\nYour guess is correct!")
                for i in range(Random_Word_len):
                    if Random_Word[i] == User_Input:
                        Word[i] = User_Input
            else:
                print("\nWrong guess!")
                Attempt -= 1
                
        except Exception:
            print("\nInvalid Input!! Please enter a valid input and one letter at a time")

    if "_" not in Word:
        print("\nCongratulations! You won the game! word is {} ".format(Random_Word))

    else:
        print("\nGame Over! the correct word is {} ".format(Random_Word))
            
    if loop:
        Play_Again = input("\nPlease enter 'yes' if you want to play again or else enter 'No' - ")
        if Play_Again.lower() not in ['yes','y']:
            loop = False
            print("Thanks for playing the word guessing game")
        
       
```
