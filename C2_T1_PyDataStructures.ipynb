```python
Jeeva_Known_Sports = {"Cricket","Football","Tennis","Running","Hockey"}
Raj_Known_Sports = {"Football","Tennis","Carrom","Chess","Swimming"}
print(Jeeva_Known_Sports)
print(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports.union(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports.intersection(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports.difference(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports.symmetric_difference(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports.issubset(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports.issuperset(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports.isdisjoint(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports = {"Cricket","Football","Tennis"}
Raj_Known_Sports = {"Cricket","Football","Tennis","Carrom","Chess","Swimming"}
Jeeva_Known_Sports.issubset(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports = {"Cricket","Football","Tennis","Carrom","Chess","Swimming"}
Raj_Known_Sports = {"Cricket","Football","Tennis"}
Jeeva_Known_Sports.issuperset(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports = {"Cricket","Football","Tennis"}
Raj_Known_Sports = {"Carrom","Chess","Swimming"}
Jeeva_Known_Sports.isdisjoint(Raj_Known_Sports)
```

```python
Jeeva_Known_Sports = {"Cricket","Football","Tennis","Running","Hockey"}
Raj_Known_Sports = {"Football","Tennis","Carrom","Chess","Swimming"}
print(Jeeva_Known_Sports)
print(Raj_Known_Sports)

print("\nBefore Intersection",Jeeva_Known_Sports)
print(Jeeva_Known_Sports.intersection(Raj_Known_Sports))
print("After Intersection",Jeeva_Known_Sports)
print("\nBefore Intersection_Update",Jeeva_Known_Sports)
Jeeva_Known_Sports.intersection_update(Raj_Known_Sports)
print("After Intersection_Update",Jeeva_Known_Sports)

Jeeva_Known_Sports = {"Cricket","Football","Tennis","Running","Hockey"}
Raj_Known_Sports = {"Football","Tennis","Carrom","Chess","Swimming"}

print("\nBefore Difference",Jeeva_Known_Sports)
print(Jeeva_Known_Sports.difference(Raj_Known_Sports))
print("After Difference",Jeeva_Known_Sports)
print("\nBefore Difference_Update",Jeeva_Known_Sports)
Jeeva_Known_Sports.difference_update(Raj_Known_Sports)
print("After Difference_Update",Jeeva_Known_Sports)

Jeeva_Known_Sports = {"Cricket","Football","Tennis","Running","Hockey"}
Raj_Known_Sports = {"Football","Tennis","Carrom","Chess","Swimming"}

print("\nBefore Symmetric_Difference",Jeeva_Known_Sports)
print(Jeeva_Known_Sports.symmetric_difference(Raj_Known_Sports))
print("After Symmetric_Difference",Jeeva_Known_Sports)
print("\nBefore Symmetric_Difference_Update",Jeeva_Known_Sports)
Jeeva_Known_Sports.symmetric_difference_update(Raj_Known_Sports)
print("After Symmetric_Difference_Update",Jeeva_Known_Sports)
```

#### Dict - Key & Value pair

```python
Stud_Marks_In_Maths = {}
print(Stud_Marks_In_Maths)
print(type(Stud_Marks_In_Maths))
```

```python
Stud_Marks_In_Maths = {"Jeeva":89, "Raj": 75, "Gugan": 95}
print(Stud_Marks_In_Maths)
print(type(Stud_Marks_In_Maths))
```

```python
Stud_Marks_In_Maths.clear()
print(Stud_Marks_In_Maths)
print(type(Stud_Marks_In_Maths))
```

```python
Stud_Maths_CopytoDB = Stud_Marks_In_Maths.copy()
print(Stud_Maths_CopytoDB)
print(type(Stud_Maths_CopytoDB))
```

```python
Stud_Marks_In_Maths.keys()
```

```python
Stud_Marks_In_Maths.values()
```

```python
Stud_Marks_In_Maths.get("Raj")
```

```python
Stud_Marks_In_Maths.get("Jeeva")
```

```python
print(Stud_Marks_In_Maths.items())
print(Stud_Maths_CopytoDB.items())
```

```python
for key,value in Stud_Marks_In_Maths.items():
    print(key,value)
```

```python
Stud_Maths_CopytoDB.update({"Somu":78})
print(Stud_Maths_CopytoDB)
```

```python
Stud_Maths_CopytoDB.update({"Somu":80})
print(Stud_Maths_CopytoDB)
```

```python
Stud_Maths_CopytoDB.popitem()
print(Stud_Maths_CopytoDB)
```

```python
try:
    Stud_Maths_CopytoDB.pop("Somu")
    print(Stud_Maths_CopytoDB)
except KeyError:
    print("Key not present")
```

```python
try:
    Stud_Maths_CopytoDB.pop("Gugan")
    print(Stud_Maths_CopytoDB)
except KeyError:
    print("Key not present")
```

```python
Stud_Maths_CopytoDB.setdefault("somu", 89)
print(Stud_Maths_CopytoDB)
```

```python
Stud_Maths_CopytoDB.setdefault("somu", 99)
print(Stud_Maths_CopytoDB)
```
