```python
Student_Results = {
    "Student Name": ["Jeeva", "Raj", "Ramu", "Somu", "Harry"],
    "Total Marks": [466, 454, 240, 387, 403],
    "Result": ["Pass", "Pass", "Fail", "Pass", "Pass"]
}

pd.DataFrame(Student_Results)
```

```python
def student_data():
    Student_Data = {
    "Student Name": ["Jeeva", "Raj", "Ramu", "Somu", "Harry"],
    "Age": [20, 21, 20, 22, 21],
    "City": ["Chennai", "Erode", "Salem", "Chennai", "Coimbatore"]
    }
    print('\n',pd.DataFrame(Student_Data))

def student_mark_data():
    Student_Mark_Data = {
    "Student Name":["Jeeva","Raj","Ramu","Somu","Harry"],
    "Tamil":[88,77,89,76,84],
    "English":[86,73,78,79,77],
    "Maths":[93,85,87,81,88],
    "Science":[94,77,93,79,83],
    "Social":[89,84,79,89,88]
    }
    print('\n',pd.DataFrame(Student_Mark_Data))

def student_results():
    Student_Results = {
    "Student Name": ["Jeeva", "Raj", "Ramu", "Somu", "Harry"],
    "Total Marks": [466, 454, 240, 387, 403],
    "Result": ["Pass", "Pass", "Fail", "Pass", "Pass"]
    }
    print('\n',pd.DataFrame(Student_Results))

User_Choice = input("""\nPlease type '1' if you want to see Student Details\n
Please type '2' if you want to see Student Marks\n
Please type '3' if you want to see Student Results """)

if User_Choice == "1":
    student_data()

elif User_Choice == "2":
    student_mark_data()

elif User_Choice == "3":
    student_results()

else:
    print("Invaild input!!")
```
