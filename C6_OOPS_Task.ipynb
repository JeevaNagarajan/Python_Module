```python
from abc import ABC, abstractmethod

# Abstraction
class ATM(ABC):

    @abstractmethod
    def login(self):
        pass
        
# Inheritance
class User(ATM):

    def __init__(self, name, atm_pin, amount):
        self.name = name
        self.__atm_pin = atm_pin # Encapsulation
        self.amount = amount

    def login(self):
        print("Hi {}, You have successfully logged into the user system".format(self.name))

    def pin_check(self, entered_pin):
        if self.__atm_pin == entered_pin:
            print("Correct PIN.")
            return True
        else:
            print("Incorrect PIN.")
            return False
            
    # Polymorphism - Method overloading 
    def withdraw_cash(self, *args):
        
        total = sum(args)

        if total <= self.amount:
            self.amount -= total
            print("Amount - {} withdrawn successfully".format(total))
            print("Remaining Balance Amount - {}".format(self.amount))
        else:
            print("Insufficient Balance")

    def exit(self):
        print("Thank you!")


# Inheritance
class Admin(User):

    def __init__(self, name):
        super().__init__(name, None, 0)

    # Polymorphism - Method overriding
    def login(self):
        print("Hi {}, You have successfully logged into the admin system".format(self.name))

    def check_atm_status(self):
        print("ATM is working properly.")

print("ATM SYSTEM\n")
print("1. User Login")
print("2. Admin Login")

choice = input("Please Enter your choice - ")

if choice == "1":

    name = input("Enter your name - ")
    atm_pin = input("Enter your PIN - ")
    amount = int(input("Enter the deposit amount - "))

    user_1 = User(name, atm_pin, amount)

    user_1.login()

    entered_pin = input("Enter the PIN again - ")

    if user_1.pin_check(entered_pin):
        # Method overloading 
        user_1.withdraw_cash(500)
        user_1.withdraw_cash(500,500)
        user_1.withdraw_cash(1000,500,500)
        
    user_1.exit()

elif choice == "2":

    name = input("Enter admin name - ")

    admin_1 = Admin(name)

    admin_1.login()
    admin_1.check_atm_status()
    admin_1.exit()

else:
    print("Invalid choice.")
```
