#### 4.Given a string, count how many vowels it has.

```python
User_Input = input("Please enter the word")
User_Input_length = len(User_Input)
#print(User_Input_length)
vowels_count = 0

for i in range(User_Input_length):
    if User_Input[i].lower() in ['a','e','i','o','u']:
        vowels_count += 1
if vowels_count != 0:       
    print("The Vowels count for the given word {} is {}".format(User_Input,vowels_count))
else:
    print("Sorry! the given word {} has no vowels on it".format(User_Input))
```

#### 5.Reverse a list without using built-in functions.

```python
Student_Id = [32,55,66,23,88]
Stud_Length = len(Student_Id)-1
Stud_Rev = []
for i in range(Stud_Length,-1,-1):
  Stud_Rev.append(Student_Id[i])

print(Stud_Rev)
```

#### 9.Replace all missing values in a DataFrame with the column’s mean.

```python
Student_Data = {
    "Student Name":["Jeeva","Raj","Ramu","Somu","Harry"],
    "Subject":["Tamil","English","Maths","Science","Social"],
    "Marks":[88,77,None,99,78]
}

Stu_DataFrame = pd.DataFrame(Student_Data)
print(Stu_DataFrame,'\n')
print(Stu_DataFrame.isna().sum(),"\n")
Stu_DataFrame.Marks.mean().astype(int)
Stu_DataFrame.Marks = Stu_DataFrame.Marks.fillna(Stu_DataFrame.Marks.mean()).astype(int)
print(Stu_DataFrame)
```

#### 10.Filter rows in a DataFrame where the "Age" column is greater than 30.

```python
Student_Data = {
    "Student Name":["Jeeva","Raj","Ramu","Somu","Harry"],
    "Subject":["Tamil","English","Maths","Science","Social"],
    "Marks":[88,77,22,99,78],
    "Age":[17,18,16,32,35]
}

Stu_DataFrame = pd.DataFrame(Student_Data)
print(Stu_DataFrame,'\n')

Stu_DataFrame_Result = Stu_DataFrame[Stu_DataFrame["Age"] > 30]
print(Stu_DataFrame_Result)
```
